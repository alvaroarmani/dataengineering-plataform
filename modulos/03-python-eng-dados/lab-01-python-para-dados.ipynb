Objetivo: praticar list/dict/set e *comprehensions* transformando um pequeno lote de pedidos — os gestos que você repete em toda ingestão.

## 1. Os dados (uma lista de dicts)

```python
pedidos = [
    {'id': 1, 'cliente': 'ana',  'categoria': 'livros',      'valor': 50.0},
    {'id': 2, 'cliente': 'bruno','categoria': 'eletronicos', 'valor': 1200.0},
    {'id': 3, 'cliente': 'ana',  'categoria': 'livros',      'valor': 30.0},
    {'id': 4, 'cliente': 'caio', 'categoria': 'eletronicos', 'valor': 800.0},
    {'id': 5, 'cliente': 'bruno','categoria': 'livros',      'valor': 45.0},
]
len(pedidos)
```

## 2. Comprehensions: filtrar e projetar
O "SELECT ... WHERE" do Python.

```python
caros = [p for p in pedidos if p['valor'] > 100]        # filtra
valores = [p['valor'] for p in pedidos]                 # projeta
print('pedidos caros:', len(caros))
print('valores:', valores)
```

## 3. set: clientes únicos (pertencimento O(1))

```python
clientes = {p['cliente'] for p in pedidos}   # set comprehension
print('clientes únicos:', clientes, '->', len(clientes))
```

## 4. dict.get: agregar receita por categoria

```python
receita = {}
for p in pedidos:
    c = p['categoria']
    receita[c] = receita.get(c, 0.0) + p['valor']
receita
```

## 5. Sua vez (mini-desafio)
Monte um **set** com as categorias distintas dos pedidos e guarde em `categorias`. Depois rode a verificação.

```python
# Complete:
categorias = {p['categoria'] for p in pedidos}
categorias
```

```python
def verificar(x):
    try:
        assert isinstance(x, set), 'categorias deve ser um set (use { } de comprehension).'
        assert x == {'livros', 'eletronicos'}, 'Deveria conter exatamente livros e eletronicos.'
        print('\u2705 Correto! Voce usou um set para achar os valores distintos.')
    except AssertionError as e:
        print('\u274c', e)

verificar(categorias)
```
