## 1. try/except específico

```python
def para_float(txt):
    try:
        return float(txt)
    except ValueError:      # captura só o erro esperado
        return None

print(para_float('3.14'), para_float('abc'))
```

## 2. Logging em vez de print

```python
import logging
logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')
log = logging.getLogger('lab')
log.info('pipeline iniciado')
log.warning('exemplo de aviso: valor inesperado')
```

## 3. Quarentena: separar válidos de inválidos

```python
linhas = ['10', '20', 'x', '5.5', 'nan?', '3']
validos, invalidos = [], []
for l in linhas:
    try:
        validos.append(float(l))
    except ValueError:
        invalidos.append(l)
        log.warning('linha invalida ignorada: %r', l)
print('validos:', validos)
print('invalidos:', invalidos)
```

## 4. Sua vez (mini-desafio)
Conte **quantas** linhas foram para a quarentena (inválidas) e guarde em `n_invalidos`. Depois verifique.

```python
# Complete:
n_invalidos = len(invalidos)
n_invalidos
```

```python
def verificar(x):
    try:
        assert x == 2, 'Havia 2 linhas invalidas ("x" e "nan?").'
        print('\u2705 Correto! Voce contou a quarentena certinho.')
    except AssertionError as e:
        print('\u274c', e)

verificar(n_invalidos)
```
