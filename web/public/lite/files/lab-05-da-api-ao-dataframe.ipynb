Objetivo: transformar um JSON aninhado (como o de uma API) em uma tabela pronta para análise.

## 1. Um payload de API (JSON aninhado)

```python
payload = {
    'pedidos': [
        {'id': 1, 'cliente': {'nome': 'ana'},   'itens': [{'valor': 50.0}, {'valor': 30.0}]},
        {'id': 2, 'cliente': {'nome': 'bruno'}, 'itens': [{'valor': 1200.0}]},
        {'id': 3, 'cliente': {'nome': 'ana'},   'itens': [{'valor': 80.0}, {'valor': 20.0}]},
    ]
}
len(payload['pedidos'])
```

## 2. Achatar: um dict plano por pedido

```python
linhas = []
for p in payload['pedidos']:
    total = sum(item['valor'] for item in p['itens'])
    linhas.append({'id': p['id'], 'cliente': p['cliente']['nome'], 'total': total})
linhas
```

## 3. Virar tabela (DataFrame) e agregar

```python
import pandas as pd
df = pd.DataFrame(linhas)
print(df)
print('\ntotal por cliente:')
df.groupby('cliente')['total'].sum().sort_values(ascending=False)
```

## 4. Sua vez (mini-desafio)
Quantos pedidos a cliente **ana** fez? Guarde em `pedidos_ana` e verifique.

```python
# Complete:
pedidos_ana = sum(1 for p in payload['pedidos'] if p['cliente']['nome'] == 'ana')
pedidos_ana
```
