Objetivo: praticar funções como valores (compor passos de um pipeline) e uma classe com estado.

## 1. Passos são funções

```python
def sem_nulos(regs):
    return [r for r in regs if r.get('valor') is not None]

def dobrar_valor(regs):
    return [{**r, 'valor': r['valor'] * 2} for r in regs]

brutos = [{'valor': 10}, {'valor': None}, {'valor': 5}]
sem_nulos(brutos)
```

## 2. Compor passos (funções como valores)

```python
def rodar(dados, passos):
    for passo in passos:
        dados = passo(dados)
    return dados

rodar(brutos, [sem_nulos, dobrar_valor])
```

## 3. Uma classe com estado

```python
class Agregador:
    def __init__(self):
        self.total = 0.0
        self.n = 0
    def adicionar(self, valor):
        self.total += valor
        self.n += 1
    def media(self):
        return self.total / self.n if self.n else 0.0

a = Agregador()
for v in [10, 20, 30]:
    a.adicionar(v)
print('total:', a.total, '| média:', a.media())
```
