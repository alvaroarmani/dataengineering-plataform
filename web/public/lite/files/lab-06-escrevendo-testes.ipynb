Objetivo: escrever uma função e **testá-la** com asserts, cobrindo casos de borda.

## 1. Uma função para testar

```python
def limpar(regs):
    'Remove registros sem valor (None).'
    return [r for r in regs if r.get('valor') is not None]

limpar([{'valor': 1}, {'valor': None}, {'valor': 2}])
```

## 2. Testes = asserts (caminho feliz + bordas)

```python
def rodar_testes():
    # caminho feliz
    assert limpar([{'valor': 1}, {'valor': None}]) == [{'valor': 1}]
    # borda: lista vazia
    assert limpar([]) == []
    # borda: nenhum nulo
    assert limpar([{'valor': 5}]) == [{'valor': 5}]
    print('\u2705 todos os testes passaram')

rodar_testes()
```

## 3. Como um teste FALHA (repare na mensagem)

```python
try:
    assert limpar([{'valor': None}]) == [{'valor': None}], 'nulos deveriam ser removidos!'
except AssertionError as e:
    print('\u274c falhou (como esperado):', e)
```

## 4. Sua vez (mini-desafio)
Implemente `soma_positivos(nums)` que soma **apenas** os números > 0. Depois rode a verificação (que age como uma mini-suíte de testes).

```python
def soma_positivos(nums):
    return sum(n for n in nums if n > 0)
```

```python
def verificar(f):
    casos = [([1, 2, 3], 6), ([-1, 2, -3], 2), ([], 0), ([-5, -2], 0)]
    try:
        for entrada, esperado in casos:
            assert f(entrada) == esperado, f'para {entrada} esperava {esperado}, veio {f(entrada)}'
        print('\u2705 Correto! Passou no caminho feliz e nas bordas.')
    except AssertionError as e:
        print('\u274c', e)

verificar(soma_positivos)
```
