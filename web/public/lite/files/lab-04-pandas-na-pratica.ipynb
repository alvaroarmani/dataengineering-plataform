## 1. Um DataFrame de pedidos

```python
import pandas as pd
import numpy as np

pedidos = pd.DataFrame({
    'cliente_id': [1, 2, 1, 3, 2],
    'categoria': ['livros', 'eletronicos', 'livros', 'casa', 'eletronicos'],
    'valor': [50.0, 1200.0, np.nan, 80.0, 800.0],
})
pedidos
```

## 2. Selecionar: coluna e máscara booleana

```python
print('coluna valor:')
print(pedidos['valor'])
print('\npedidos caros (>100):')
pedidos[pedidos['valor'] > 100]
```

## 3. Tratar NaN e agregar com groupby

```python
limpo = pedidos.dropna(subset=['valor'])          # remove a linha sem valor
receita = limpo.groupby('categoria')['valor'].sum().sort_values(ascending=False)
receita
```

## 4. Juntar tabelas com merge

```python
clientes = pd.DataFrame({'cliente_id': [1, 2, 3], 'nome': ['ana', 'bruno', 'caio']})
pedidos.merge(clientes, on='cliente_id', how='left')[['nome', 'categoria', 'valor']]
```

## 5. Sua vez (mini-desafio)
Calcule a **receita total** (ignorando `NaN`) e guarde em `receita_total`. Depois verifique.

```python
# Complete:
receita_total = pedidos['valor'].sum()   # .sum() já ignora NaN por padrão
receita_total
```

```python
def verificar(x):
    try:
        assert abs(float(x) - 2130.0) < 1e-6, 'A receita total (ignorando NaN) deveria ser 2130.'
        print('\u2705 Correto! .sum() ignora NaN e somou 50+1200+80+800 = 2130.')
    except AssertionError as e:
        print('\u274c', e)

verificar(receita_total)
```
